```python
def second_max(arr):

    if len(arr) < 2:
        return None

    max_val = arr[0]
    for i in range(1, len(arr)):
        if arr[i] > max_val:
            max_val = arr[i]

    min_diff = float("inf")
    result = None

    for i in range(len(arr)):
        diff = max_val - arr[i]
        if diff > 0 and diff < min_diff:
            min_diff = diff
            result = arr[i]  

    return result


# Function calls
print(second_max([1, 2, 3, 4, 5])) 
print(second_max([12, 1, 45, 22, 31, 60, 0]))  
print(second_max([1,1,1,1,1]))
```
